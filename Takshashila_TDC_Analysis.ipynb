```python

# STEP 8: DETAILED COST INVESTIGATION

import pandas as pd
import numpy as np

# 1. Identify available cost columns
cost_columns = [
    'Primaryfrt_amount',
    'secfrt_amount',
    'Hand_value',
    'TDC(INCL TU COST)',
    'TU/SATNA',
    'FrtAmt TP',
    'Route Frt Amount',
    'Primary FRT',
    'Frt Marze',
    'Primary FRT2',
    'Sec Frt',
    'Sec_frt_acc',
    'IGST Amount',
    'SGST Amount',
    'CGST Amount'
]

available_costs = [
    col for col in cost_columns
    if col in df.columns
]

# 2. Convert cost columns to numeric
for col in available_costs:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 3. Create cost audit table
cost_audit = pd.DataFrame({
    'Column': available_costs,
    'Non_Missing': [df[col].count() for col in available_costs],
    'Missing': [df[col].isna().sum() for col in available_costs],
    'Total': [df[col].sum() for col in available_costs],
    'Mean': [df[col].mean() for col in available_costs],
    'Maximum': [df[col].max() for col in available_costs]
})

print("COST AUDIT REPORT")
display(cost_audit.round(2))

# 4. TDC distribution
print("\nTDC DISTRIBUTION")

display(
    df['TDC(INCL TU COST)'].describe(
        percentiles=[0.50, 0.75, 0.90, 0.95, 0.99]
    ).to_frame()
)

# 5. Top 15 TDC transactions
print("\nTOP 15 TDC TRANSACTIONS")

top_tdc = df.nlargest(
    15,
    'TDC(INCL TU COST)'
)

top_columns = [
    'Plant',
    'Qty',
    'Primaryfrt_amount',
    'secfrt_amount',
    'Hand_value',
    'TDC(INCL TU COST)'
]

top_columns = [
    col for col in top_columns
    if col in df.columns
]

display(top_tdc[top_columns].round(2))

# 6. Save the cost audit
cost_audit.to_csv(
    '/content/TDC_Cost_Audit.csv',
    index=False
)

print("\nSUCCESS: COST INVESTIGATION COMPLETED")
```

```python

# STEP 9: TRANSACTION-LEVEL TDC RECONCILIATION

import pandas as pd
import numpy as np

# 1. Define required columns
required = [
    'Plant',
    'Qty',
    'Primaryfrt_amount',
    'secfrt_amount',
    'Hand_value',
    'TDC(INCL TU COST)'
]

# Check columns
missing_cols = [c for c in required if c not in df.columns]

if missing_cols:
    raise ValueError(f"Missing columns: {missing_cols}")

# 2. Create a separate analysis dataframe
audit = df[required].copy()

# Convert numerical fields
numeric_cols = [
    'Qty',
    'Primaryfrt_amount',
    'secfrt_amount',
    'Hand_value',
    'TDC(INCL TU COST)'
]

for col in numeric_cols:
    audit[col] = pd.to_numeric(
        audit[col], errors='coerce'
    )

# 3. Identify missing cost components
components = [
    'Primaryfrt_amount',
    'secfrt_amount',
    'Hand_value'
]

audit['Missing_Component_Count'] = (
    audit[components].isna().sum(axis=1)
)

audit['Complete_Record'] = (
    audit['Missing_Component_Count'] == 0
)

# 4. Calculate exploratory component total
audit['Component_Total'] = (
    audit[components].fillna(0).sum(axis=1)
)

# 5. Calculate reconciliation gap
audit['TDC_Gap'] = (
    audit['TDC(INCL TU COST)'] -
    audit['Component_Total']
)

# 6. Overall reconciliation summary
print("===== RECONCILIATION SUMMARY =====")

print("Total records:", len(audit))

print("Complete records:",
      audit['Complete_Record'].sum())

print("Incomplete records:",
      (~audit['Complete_Record']).sum())

print("Reported TDC:",
      audit['TDC(INCL TU COST)'].sum())

print("Component total:",
      audit['Component_Total'].sum())

print("Exploratory TDC gap:",
      audit['TDC_Gap'].sum())

# 7. Top 20 positive gaps
print("\n===== TOP 20 POSITIVE TDC GAPS =====")

top_gaps = audit.nlargest(20, 'TDC_Gap')

display(top_gaps.round(2))

# 8. Plant-wise reconciliation
print("\n===== PLANT-WISE TDC GAP =====")

plant_gap = audit.groupby('Plant', dropna=False).agg(
    Total_Tonnes=('Qty', 'sum'),
    Reported_TDC=('TDC(INCL TU COST)', 'sum'),
    Component_Cost=('Component_Total', 'sum'),
    TDC_Gap=('TDC_Gap', 'sum'),
    Records=('Plant', 'size')
).reset_index()

plant_gap['Gap_Per_Tonne'] = (
    plant_gap['TDC_Gap'] /
    plant_gap['Total_Tonnes'].replace(0, np.nan)
)

display(
    plant_gap.sort_values(
        'TDC_Gap',
        ascending=False
    ).head(15).round(2)
)

print("\nSTEP 9 COMPLETED SUCCESSFULLY")
```

```python

# STEP 10: VALIDATED TDC RECONCILIATION

import pandas as pd
import numpy as np

components = [
    'Primaryfrt_amount',
    'secfrt_amount',
    'Hand_value'
]

# Work on a separate copy
rec = df.copy()

# Identify records with all three cost components
rec['Complete_Cost_Record'] = (
    rec[components].notna().all(axis=1)
)

# Calculate component cost only for complete records
rec['Component_Total'] = rec[components].sum(
    axis=1,
    min_count=3
)

# Calculate gap only for complete records
rec['TDC_Gap'] = (
    rec['TDC(INCL TU COST)'] -
    rec['Component_Total']
)

# 1. Complete versus incomplete summary
print("===== DATA COVERAGE =====")

print("Total records:", len(rec))
print("Complete records:",
      rec['Complete_Cost_Record'].sum())
print("Incomplete records:",
      (~rec['Complete_Cost_Record']).sum())

# 2. Reconciliation for complete records only
complete = rec[rec['Complete_Cost_Record']].copy()

print("\n===== COMPLETE RECORD RECONCILIATION =====")

print("Reported TDC:",
      complete['TDC(INCL TU COST)'].sum())

print("Three component cost:",
      complete['Component_Total'].sum())

print("Reconciliation gap:",
      complete['TDC_Gap'].sum())

print("Gap per tonne:",
      complete['TDC_Gap'].sum() /
      complete['Qty'].sum())

# 3. Gap distribution
print("\n===== GAP DISTRIBUTION =====")

display(complete['TDC_Gap'].describe())

# 4. Largest positive gaps
print("\n===== TOP 20 COMPLETE RECORD GAPS =====")

display(
    complete.nlargest(20, 'TDC_Gap')[
        [
            'Plant',
            'Qty',
            'Primaryfrt_amount',
            'secfrt_amount',
            'Hand_value',
            'TDC(INCL TU COST)',
            'Component_Total',
            'TDC_Gap'
        ]
    ].round(2)
)

# 5. Plant-wise analysis
print("\n===== PLANT-WISE COMPLETE RECORD ANALYSIS =====")

plant_analysis = complete.groupby('Plant').agg(
    Tonnes=('Qty', 'sum'),
    Reported_TDC=('TDC(INCL TU COST)', 'sum'),
    Component_Cost=('Component_Total', 'sum'),
    Gap=('TDC_Gap', 'sum'),
    Records=('Plant', 'size')
).reset_index()

plant_analysis['Gap_Per_Tonne'] = (
    plant_analysis['Gap'] /
    plant_analysis['Tonnes'].replace(0, np.nan)
)

display(
    plant_analysis.sort_values(
        'Gap',
        ascending=False
    ).head(15).round(2)
)

print("\nVALIDATED RECONCILIATION COMPLETED")
```

```python

# STEP 11: INVESTIGATE TRANSFER UNIT COST

import pandas as pd
import numpy as np

# Create analysis copy
tu = df.copy()

components = [
    'Primaryfrt_amount',
    'secfrt_amount',
    'Hand_value'
]

numeric_cols = components + [
    'TDC(INCL TU COST)',
    'Qty'
]

for col in numeric_cols:
    tu[col] = pd.to_numeric(tu[col], errors='coerce')

# Calculate component total only on complete records
tu['Complete'] = tu[components].notna().all(axis=1)

tu['Component_Total'] = tu[components].sum(
    axis=1, min_count=3
)

complete_tu = tu[tu['Complete']].copy()

complete_tu['TDC_Gap'] = (
    complete_tu['TDC(INCL TU COST)'] -
    complete_tu['Component_Total']
)

# 1. Check transfer unit categories
print("TRANSFER UNIT VALUES")

display(
    tu['TU/SATNA'].value_counts(dropna=False)
)

# 2. Group-level reconciliation
print("\nTRANSFER UNIT COST RECONCILIATION")

tu_analysis = complete_tu.groupby(
    'TU/SATNA', dropna=False
).agg(
    Records=('Qty', 'size'),
    Tonnes=('Qty', 'sum'),
    Reported_TDC=('TDC(INCL TU COST)', 'sum'),
    Component_Cost=('Component_Total', 'sum'),
    TDC_Gap=('TDC_Gap', 'sum')
).reset_index()

tu_analysis['Gap_Per_Tonne'] = (
    tu_analysis['TDC_Gap'] /
    tu_analysis['Tonnes'].replace(0, np.nan)
)

display(tu_analysis.round(2))

# 3. Compare UNIT-B with other sourcing units
print("\nUNIT-B VS OTHER UNITS")

complete_tu['Unit_Group'] = np.where(
    complete_tu['TU/SATNA'].astype(str).str.upper().str.strip()
    == 'UNIT-B',
    'UNIT-B',
    'OTHER / MISSING'
)

comparison = complete_tu.groupby('Unit_Group').agg(
    Records=('Qty', 'size'),
    Tonnes=('Qty', 'sum'),
    Reported_TDC=('TDC(INCL TU COST)', 'sum'),
    Component_Cost=('Component_Total', 'sum'),
    TDC_Gap=('TDC_Gap', 'sum')
).reset_index()

comparison['Gap_Per_Tonne'] = (
    comparison['TDC_Gap'] /
    comparison['Tonnes'].replace(0, np.nan)
)

display(comparison.round(2))

print("\nSTEP 11 COMPLETED")
```

```python

# STEP 14: SECTION A1 - CEO NETWORK SNAPSHOT

import pandas as pd
import numpy as np

# Core business columns
required = [
    'DATE',
    'Plant',
    'Sales Office1',
    'Dir Plant',
    'CUSTOMER CODE',
    'Qty',
    'TDC(INCL TU COST)',
    'Primaryfrt_amount',
    'secfrt_amount',
    'Hand_value',
    'RAKE/ROAD',
    'Grade2'
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(f"Missing columns: {missing}")

# Create separate snapshot dataframe
snap = df[required].copy()

# Convert numeric columns
numeric_cols = [
    'Qty',
    'TDC(INCL TU COST)',
    'Primaryfrt_amount',
    'secfrt_amount',
    'Hand_value'
]

for col in numeric_cols:
    snap[col] = pd.to_numeric(snap[col], errors='coerce')

# Parse transaction dates
snap['DATE'] = pd.to_datetime(
    snap['DATE'], errors='coerce', dayfirst=True
)

# Network KPIs
total_records = len(snap)
total_tonnes = snap['Qty'].sum()
total_tdc = snap['TDC(INCL TU COST)'].sum()

kpis = {
    'Total Records': total_records,
    'Total Volume (Tonnes)': total_tonnes,
    'Reported TDC (INR)': total_tdc,
    'TDC per Tonne (INR)': total_tdc / total_tonnes,
    'Unique Supply Points': snap['Plant'].nunique(),
    'Unique Sales Offices': snap['Sales Office1'].nunique(),
    'Unique Customers': snap['CUSTOMER CODE'].nunique(),
    'Unique Product Families': snap['Grade2'].nunique(),
    'Reporting Start Date': snap['DATE'].min(),
    'Reporting End Date': snap['DATE'].max()
}

print("========== CEO NETWORK SNAPSHOT ==========")

for metric, value in kpis.items():
    print(metric, ":", value)

# Cost component overview
print("\n========== COST COMPONENTS ==========")

for col in [
    'Primaryfrt_amount',
    'secfrt_amount',
    'Hand_value'
]:
    print(
        col,
        "Total INR:", snap[col].sum(),
        "Missing:", snap[col].isna().sum()
    )

# Depot vs direct records and volume
print("\n========== DEPOT VS DIRECT ==========")

distribution = snap.groupby(
    'Dir Plant', dropna=False
).agg(
    Records=('Qty', 'size'),
    Tonnes=('Qty', 'sum'),
    TDC=('TDC(INCL TU COST)', 'sum')
).reset_index()

distribution['Volume_Share_Percent'] = (
    distribution['Tonnes'] / total_tonnes * 100
)

distribution['TDC_Per_Tonne'] = (
    distribution['TDC'] /
    distribution['Tonnes'].replace(0, np.nan)
)

display(distribution.round(2))

# Primary mode summary
print("\n========== PRIMARY MODE ==========")

mode_summary = snap.groupby(
    'RAKE/ROAD', dropna=False
).agg(
    Records=('Qty', 'size'),
    Tonnes=('Qty', 'sum'),
    TDC=('TDC(INCL TU COST)', 'sum')
).reset_index()

mode_summary['Volume_Share_Percent'] = (
    mode_summary['Tonnes'] / total_tonnes * 100
)

display(mode_summary.round(2))

# Save outputs
kpi_df = pd.DataFrame(
    list(kpis.items()),
    columns=['Metric', 'Value']
)

kpi_df.to_csv(
    '/content/Section_A1_CEO_Snapshot.csv',
    index=False
)

distribution.to_csv(
    '/content/Depot_Direct_Summary.csv',
    index=False
)

mode_summary.to_csv(
    '/content/Primary_Mode_Summary.csv',
    index=False
)

print("\nSECTION A1 INITIAL ANALYSIS COMPLETED")
```

```python

# STEP 16: SECTION B1 - VOLUME AND TDC ANALYSIS

import pandas as pd
import numpy as np

# Prepare analysis data
b1 = df.copy()

b1['Qty'] = pd.to_numeric(b1['Qty'], errors='coerce')

b1['TDC(INCL TU COST)'] = pd.to_numeric(
    b1['TDC(INCL TU COST)'], errors='coerce'
)

total_volume = b1['Qty'].sum()

# FUNCTION TO GENERATE NETWORK SUMMARY
def network_summary(data, group_column):

    result = data.groupby(
        group_column,
        dropna=False
    ).agg(
        Records=('Qty', 'size'),
        Tonnes=('Qty', 'sum'),
        Total_TDC=('TDC(INCL TU COST)', 'sum')
    ).reset_index()

    result['Volume_Share_%'] = (
        result['Tonnes'] / total_volume * 100
    )

    result['TDC_Per_Tonne'] = (
        result['Total_TDC'] /
        result['Tonnes'].replace(0, np.nan)
    )

    result['TDC_Share_%'] = (
        result['Total_TDC'] /
        b1['TDC(INCL TU COST)'].sum() * 100
    )

    return result.round(2)


# 1. RAIL VS ROAD
print("========== B1: RAIL VS ROAD ==========")

rail_road = network_summary(
    b1,
    'RAKE/ROAD'
)

display(rail_road)


# 2. DEPOT VS DIRECT
print("\n========== B1: DEPOT VS DIRECT ==========")

depot_direct = network_summary(
    b1,
    'Dir Plant'
)

display(depot_direct)


# 3. Export results
rail_road.to_csv(
    '/content/B1_Rail_Road_Analysis.csv',
    index=False
)

depot_direct.to_csv(
    '/content/B1_Depot_Direct_Analysis.csv',
    index=False
)

print("\nSECTION B1 COMPLETED")
```

```python

# STEP 17: SECTION B2 - VEHICLE AND PRODUCT MIX

import pandas as pd
import numpy as np

b2 = df.copy()

b2['Qty'] = pd.to_numeric(b2['Qty'], errors='coerce')
b2['TDC(INCL TU COST)'] = pd.to_numeric(
    b2['TDC(INCL TU COST)'], errors='coerce'
)

total_tonnes = b2['Qty'].sum()

# Function for volume mix analysis
def mix_analysis(data, column):

    result = data.groupby(
        column, dropna=False
    ).agg(
        Records=('Qty', 'size'),
        Tonnes=('Qty', 'sum'),
        Total_TDC=('TDC(INCL TU COST)', 'sum')
    ).reset_index()

    result['Volume_Share_%'] = (
        result['Tonnes'] / total_tonnes * 100
    )

    result['TDC_Per_Tonne'] = (
        result['Total_TDC'] /
        result['Tonnes'].replace(0, np.nan)
    )

    return result.sort_values(
        'Tonnes', ascending=False
    ).round(2)


# 1. Vehicle / movement class
print("========== VEHICLE / MOVEMENT CLASS ==========")

vehicle_mix = mix_analysis(
    b2, 'Truck Type Clean'
)

display(vehicle_mix)

# 2. Product family
print("\n========== PRODUCT FAMILY MIX ==========")

product_mix = mix_analysis(
    b2, 'Grade2'
)

display(product_mix)

# 3. Save results
vehicle_mix.to_csv(
    '/content/B2_Vehicle_Mix.csv',
    index=False
)

product_mix.to_csv(
    '/content/B2_Product_Mix.csv',
    index=False
)

print("\nSECTION B2 COMPLETED")
```

```python

# STEP 18: SECTION B3 - CUSTOMER CONCENTRATION

import pandas as pd
import numpy as np

# Prepare customer data
customer_df = df.copy()

customer_df['Qty'] = pd.to_numeric(
    customer_df['Qty'], errors='coerce'
)

customer_df['TDC(INCL TU COST)'] = pd.to_numeric(
    customer_df['TDC(INCL TU COST)'], errors='coerce'
)

total_volume = customer_df['Qty'].sum()

# Aggregate customer-level information
customer_summary = customer_df.groupby(
    'CUSTOMER CODE'
).agg(
    Records=('Qty', 'size'),
    Total_Tonnes=('Qty', 'sum'),
    Total_TDC=('TDC(INCL TU COST)', 'sum')
).reset_index()

# Rank customers by volume
customer_summary = customer_summary.sort_values(
    'Total_Tonnes',
    ascending=False
).reset_index(drop=True)

customer_summary['Rank'] = (
    customer_summary.index + 1
)

# Customer volume share
customer_summary['Volume_Share_%'] = (
    customer_summary['Total_Tonnes'] /
    total_volume * 100
)

# Cumulative share
customer_summary['Cumulative_Share_%'] = (
    customer_summary['Volume_Share_%'].cumsum()
)

# TDC per tonne
customer_summary['TDC_Per_Tonne'] = (
    customer_summary['Total_TDC'] /
    customer_summary['Total_Tonnes'].replace(0, np.nan)
)

# 1. Top 10 customers
print("========== TOP 10 CUSTOMERS ==========")

display(
    customer_summary.head(10).round(2)
)

# 2. Concentration summary
print("\n========== CUSTOMER CONCENTRATION ==========")

concentration = []

for n in [10, 20, 50, 100]:
    share = customer_summary.head(n)[
        'Total_Tonnes'
    ].sum() / total_volume * 100

    concentration.append({
        'Top_Customers': n,
        'Volume_Share_%': round(share, 2)
    })

concentration_df = pd.DataFrame(concentration)

display(concentration_df)

# 3. HHI calculation
customer_shares = (
    customer_summary['Volume_Share_%'] / 100
)

hhi = (customer_shares ** 2).sum() * 10000

print("\nTotal unique customers:",
      customer_summary['CUSTOMER CODE'].nunique())

print("Customer HHI:", round(hhi, 2))

# 4. Export results
customer_summary.to_csv(
    '/content/B3_Customer_Ranking.csv',
    index=False
)

concentration_df.to_csv(
    '/content/B3_Customer_Concentration.csv',
    index=False
)

print("\nSECTION B3 COMPLETED")
```

```python

# STEP 19: SECTION B4 - TDC COST BREAKDOWN

import pandas as pd
import numpy as np

b4 = df.copy()

total_tonnes = pd.to_numeric(
    b4['Qty'], errors='coerce'
).sum()

total_tdc = pd.to_numeric(
    b4['TDC(INCL TU COST)'], errors='coerce'
).sum()

cost_columns = {
    'Primary Freight': 'Primaryfrt_amount',
    'Secondary Freight': 'secfrt_amount',
    'Handling': 'Hand_value'
}

results = []

# Calculate each component
for component, column in cost_columns.items():

    amount = pd.to_numeric(
        b4[column], errors='coerce'
    ).sum()

    missing = b4[column].isna().sum()

    results.append({
        'Component': component,
        'Total_Cost_INR': amount,
        'Share_of_TDC_%': amount / total_tdc * 100,
        'Cost_Per_Tonne_INR': amount / total_tonnes,
        'Missing_Records': missing
    })

# Calculate unreconciled residual
component_total = sum(
    row['Total_Cost_INR'] for row in results
)

residual = total_tdc - component_total

results.append({
    'Component': 'Unreconciled Residual',
    'Total_Cost_INR': residual,
    'Share_of_TDC_%': residual / total_tdc * 100,
    'Cost_Per_Tonne_INR': residual / total_tonnes,
    'Missing_Records': np.nan
})

# Create final table
tdc_breakdown = pd.DataFrame(results)

print("========== B4: TDC BREAKDOWN ==========")

display(tdc_breakdown.round(2))

# Validation
print("\n========== VALIDATION ==========")

print("Total volume:", round(total_tonnes, 2))
print("Reported TDC:", round(total_tdc, 2))

print("Sum of component costs + residual:",
      round(
          tdc_breakdown['Total_Cost_INR'].sum(),
          2
      )
)

print("Total TDC per tonne:",
      round(total_tdc / total_tonnes, 2)
)

# Export
tdc_breakdown.to_csv(
    '/content/B4_TDC_Breakdown.csv',
    index=False
)

print("\nSECTION B4 COMPLETED")
```

```python
# C2: Primary freight rate per tonne-km

c2 = df[
    df["Dir Plant"].astype(str).str.upper().str.strip() == "DEPOT"
].copy()

for col in ["Qty", "Primary distance", "Primaryfrt_amount"]:
    c2[col] = pd.to_numeric(c2[col], errors="coerce")

# Valid primary movement records
c2 = c2[
    (c2["Qty"] > 0) &
    (c2["Primary distance"] > 0) &
    (c2["Primaryfrt_amount"].notna())
].copy()

c2["Primary_tonne_km"] = c2["Qty"] * c2["Primary distance"]

# Overall weighted rate
overall = (
    c2["Primaryfrt_amount"].sum() /
    c2["Primary_tonne_km"].sum()
)

# Mode-wise weighted rates
mode_summary = c2.groupby("RAKE/ROAD").apply(
    lambda g: pd.Series({
        "Records": len(g),
        "Volume_Tonnes": g["Qty"].sum(),
        "Primary_Freight_Rs": g["Primaryfrt_amount"].sum(),
        "Tonne_Km": g["Primary_tonne_km"].sum(),
        "Rate_Rs_per_Tonne_Km":
            g["Primaryfrt_amount"].sum() /
            g["Primary_tonne_km"].sum()
    }),
    include_groups=False
)

print("C2 — Primary Freight Rate")
print("Overall rate:", round(overall, 2))
print(mode_summary.round(2))
```

```python
# C3 — Depot-routed end-mile distance analysis

import pandas as pd
import numpy as np

# Select depot-routed movements
c3 = df[
    df["Dir Plant"].astype(str).str.upper().str.strip() == "DEPOT"
].copy()

# Convert columns to numeric
for col in ["Qty", "Distance", "secfrt_amount"]:
    c3[col] = pd.to_numeric(c3[col], errors="coerce")

# Keep valid records; retain zero-freight rows
c3 = c3[
    (c3["Qty"] > 0) &
    (c3["Distance"] > 0) &
    (c3["secfrt_amount"].notna())
].copy()

# Create distance bands
conditions = [
    c3["Distance"] < 20,
    (c3["Distance"] >= 20) & (c3["Distance"] <= 30),
    (c3["Distance"] > 30) & (c3["Distance"] <= 40),
    c3["Distance"] > 40
]

labels = ["<20 km", "20–30 km", "30–40 km", ">40 km"]

c3["Distance_Band"] = np.select(
    conditions,
    labels,
    default="Unclassified"
)

# Calculate tonne-km
c3["Tonne_km"] = c3["Qty"] * c3["Distance"]

# Summarise each distance band
summary = c3.groupby("Distance_Band", observed=True).agg(
    Records=("Qty", "count"),
    Volume_Tonnes=("Qty", "sum"),
    Secondary_Freight_Rs=("secfrt_amount", "sum"),
    Tonne_Km=("Tonne_km", "sum")
).reindex(labels)

# Calculate required metrics
total_volume = c3["Qty"].sum()

summary["Volume_Share_%"] = (
    summary["Volume_Tonnes"] / total_volume * 100
)

summary["Rate_Rs_per_Tonne_Km"] = (
    summary["Secondary_Freight_Rs"] / summary["Tonne_Km"]
)

summary["Cost_Rs_per_Tonne"] = (
    summary["Secondary_Freight_Rs"] / summary["Volume_Tonnes"]
)

print("C3 — DEPOT-ROUTED END-MILE ANALYSIS")
print("=" * 65)
print("Valid records:", f"{len(c3):,}")
print("Total valid volume:", f"{total_volume:,.2f} tonnes")
print()
print(summary.round(2).to_string())

# Additional diagnostic: zero-freight records
zero_freight = c3[c3["secfrt_amount"] == 0]

print("\nZERO SECONDARY FREIGHT DIAGNOSTIC")
print("Zero-freight records:", f"{len(zero_freight):,}")
print("Zero-freight volume:", f"{zero_freight['Qty'].sum():,.2f} tonnes")
```

```python
# C4 — Zero secondary freight analysis
# Using the actual project column: Supplier Point

import pandas as pd
import numpy as np

c4 = df[
    df["Dir Plant"].astype(str).str.strip().str.upper() == "DEPOT"
].copy()

c4["Qty"] = pd.to_numeric(c4["Qty"], errors="coerce")
c4["secfrt_amount"] = pd.to_numeric(
    c4["secfrt_amount"], errors="coerce"
)

# Include all depot-routed records with valid positive volume
c4 = c4[c4["Qty"] > 0].copy()

# Zero secondary freight records
zero = c4[c4["secfrt_amount"] == 0].copy()

print("C4 — ZERO SECONDARY FREIGHT")
print("Records:", len(zero))
print("Volume:", round(zero["Qty"].sum(), 2))

# Supplier point analysis
summary = c4.groupby("Supplier Point").agg(
    Total_Records=("Qty", "count"),
    Total_Volume=("Qty", "sum")
)

zero_summary = zero.groupby("Supplier Point").agg(
    Zero_Records=("Qty", "count"),
    Zero_Volume=("Qty", "sum")
)

result = summary.join(zero_summary, how="left").fillna(0)

result["Zero_Volume_Share_%"] = (
    result["Zero_Volume"] /
    result["Total_Volume"] * 100
)

# Identify supplier points where 100% volume has zero freight
all_zero = result[
    np.isclose(result["Zero_Volume_Share_%"], 100, atol=0.01)
].sort_values("Total_Volume", ascending=False)

print("\nSUPPLIER POINTS WITH 100% ZERO FREIGHT")
print(all_zero.round(2).to_string())

print("\nTOP 10 SUPPLIER POINTS BY ZERO-FREIGHT VOLUME")
print(
    result.sort_values("Zero_Volume", ascending=False)
    .head(10).round(2).to_string()
)
```

```python
# D1 — Supply point secondary cost ranking

import pandas as pd
import numpy as np

# 1. Select depot-routed movements
d1 = df[
    df["Dir Plant"].astype(str).str.strip().str.upper() == "DEPOT"
].copy()

# 2. Convert required fields
for col in ["Qty", "Distance", "secfrt_amount"]:
    d1[col] = pd.to_numeric(d1[col], errors="coerce")

# 3. Keep positive-volume records
d1 = d1[d1["Qty"] > 0].copy()

# 4. Calculate total volume by supplier point
volume_summary = d1.groupby("Supplier Point").agg(
    Total_Volume=("Qty", "sum"),
    Total_Records=("Qty", "count")
)

# 5. Identify records with reported freight
reported = d1[d1["secfrt_amount"].notna()].copy()

reported["Reported_Tonne_Km"] = (
    reported["Qty"] * reported["Distance"]
)

# 6. Aggregate reported freight and corresponding volume
cost_summary = reported.groupby("Supplier Point").agg(
    Reported_Freight=("secfrt_amount", "sum"),
    Reported_Volume=("Qty", "sum"),
    Reported_Records=("Qty", "count")
)

# 7. Calculate distance and tonne-km using valid distances
distance_valid = reported[
    (reported["Distance"] > 0)
].copy()

distance_summary = distance_valid.groupby("Supplier Point").agg(
    Tonne_Km=("Reported_Tonne_Km", "sum"),
    Distance_Volume=("Qty", "sum"),
    Weighted_Distance_Numerator=(
        "Distance",
        lambda x: (
            x * distance_valid.loc[x.index, "Qty"]
        ).sum()
    )
)

# 8. Combine results
result = volume_summary.join(cost_summary, how="left")
result = result.join(distance_summary, how="left")

result = result.fillna(0)

# 9. Calculate metrics
result["Freight_Coverage_%"] = (
    result["Reported_Volume"] /
    result["Total_Volume"] * 100
)

result["Secondary_Cost_Per_Tonne"] = np.where(
    result["Reported_Volume"] > 0,
    result["Reported_Freight"] / result["Reported_Volume"],
    np.nan
)

result["Secondary_Rate_Per_Tonne_Km"] = np.where(
    result["Tonne_Km"] > 0,
    result["Reported_Freight"] / result["Tonne_Km"],
    np.nan
)

result["Weighted_Avg_Distance"] = np.where(
    result["Distance_Volume"] > 0,
    result["Weighted_Distance_Numerator"] /
    result["Distance_Volume"],
    np.nan
)

result["Missing_Freight_Volume"] = (
    result["Total_Volume"] - result["Reported_Volume"]
)

# 10. Apply assessment eligibility threshold
eligible = result[
    result["Total_Volume"] >= 10000
].copy()

# 11. Rank five highest reported secondary cost per tonne
eligible = eligible.sort_values(
    "Secondary_Cost_Per_Tonne",
    ascending=False
)

print("D1 — TOP 5 HIGHEST-COST SUPPLIER POINTS")
print("=" * 90)

columns = [
    "Total_Records",
    "Total_Volume",
    "Reported_Freight",
    "Secondary_Cost_Per_Tonne",
    "Weighted_Avg_Distance",
    "Secondary_Rate_Per_Tonne_Km",
    "Freight_Coverage_%",
    "Missing_Freight_Volume"
]

print(eligible[columns].head(5).round(2).to_string())

print("\nOVERALL ELIGIBLE SUPPLIER POINTS")
print("Count:", len(eligible))

print("\nELIGIBLE SUPPLIER POINTS — COMPLETE RANKING")
print(
    eligible[columns].round(2).to_string()
)

# 12. Benchmark against all eligible supplier points
print("\nNETWORK BENCHMARK")
print("Median cost per tonne:",
      round(eligible["Secondary_Cost_Per_Tonne"].median(), 2))

print("Median weighted distance:",
      round(eligible["Weighted_Avg_Distance"].median(), 2))

print("Median rate per tonne-km:",
      round(eligible["Secondary_Rate_Per_Tonne_Km"].median(), 2))
```

```python
# D2 — Annual secondary freight savings from reducing >40 km leads to 30 km

import pandas as pd
import numpy as np

d2 = df[
    df["Dir Plant"].astype(str).str.strip().str.upper() == "DEPOT"
].copy()

for col in ["Qty", "Distance", "secfrt_amount"]:
    d2[col] = pd.to_numeric(d2[col], errors="coerce")

# Valid records above 40 km
long_block = d2[
    (d2["Qty"] > 0) &
    (d2["Distance"] > 40) &
    (d2["secfrt_amount"].notna())
].copy()

# Separate freight-paying and zero-freight records
paid = long_block[long_block["secfrt_amount"] > 0].copy()
zero = long_block[long_block["secfrt_amount"] == 0].copy()

# Calculate tonne-km for freight-paying records
paid["Tonne_Km"] = paid["Qty"] * paid["Distance"]

# Current position
volume = paid["Qty"].sum()
freight = paid["secfrt_amount"].sum()
tonne_km = paid["Tonne_Km"].sum()

current_rate = freight / tonne_km
weighted_distance = tonne_km / volume

# Estimate cost if all freight-paying volume is delivered at 30 km
target_distance = 30
target_tonne_km = volume * target_distance
estimated_new_freight = target_tonne_km * current_rate

# Quarterly and annual savings
quarterly_saving = freight - estimated_new_freight
annual_saving = quarterly_saving * 4

print("D2 — SAVINGS FROM REDUCING >40 KM TO 30 KM")
print("=" * 65)

print("\nFREIGHT-PAYING LONG-DISTANCE MOVEMENTS")
print("Records:", f"{len(paid):,}")
print("Volume:", f"{volume:,.2f} tonnes")
print("Current freight: ₹", f"{freight:,.2f}")
print("Current tonne-km:", f"{tonne_km:,.2f}")
print("Weighted average distance:", round(weighted_distance, 2), "km")
print("Weighted freight rate: ₹", round(current_rate, 2), "/tonne-km")

print("\n30 KM SCENARIO")
print("Target tonne-km:", f"{target_tonne_km:,.2f}")
print("Estimated freight at 30 km: ₹", f"{estimated_new_freight:,.2f}")

print("\nESTIMATED SAVINGS")
print("Quarterly saving: ₹", f"{quarterly_saving:,.2f}")
print("Annual saving (x4): ₹", f"{annual_saving:,.2f}")
print("Annual saving in crore: ₹", round(annual_saving / 1e7, 2), "Cr")

print("\nZERO-FREIGHT LONG-DISTANCE MOVEMENTS — EXCLUDED")
print("Records:", f"{len(zero):,}")
print("Volume:", f"{zero['Qty'].sum():,.2f} tonnes")

print("\nASSUMPTION")
print("Same observed freight rate per tonne-km is maintained.")
print("April–June 2026 quarter is annualised by multiplying by 4.")
```
